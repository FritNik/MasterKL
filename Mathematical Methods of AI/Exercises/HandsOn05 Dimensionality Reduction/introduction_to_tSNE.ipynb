# tSNE

In this notebook, we will instead use nonlinear dimensionality reduction, and namely tSNE

````python
iris = load_iris(as_frame=False)
X = np.array(iris.data)
print(X.shape)
y = iris.target
````

For simplicity, we will use the `sklearn` function.
Compare the results obtained with the PCA results.

Experiment with various `perplexity` values and observe the results

````python
from sklearn.manifold import TSNE
import matplotlib.pyplot as plt

# t-SNE Embedding berechnen
tsne = TSNE(n_components=2, perplexity=10, learning_rate='auto', init='random')
X_embedded = tsne.fit_transform(X)

# Scatterplot
plt.figure(figsize=(6,5))
plt.scatter(X_embedded[:, 0], X_embedded[:, 1], c=y, cmap='tab10')
plt.xlabel("t-SNE Dimension 1")
plt.ylabel("t-SNE Dimension 2")
plt.title("t-SNE embedding")
plt.colorbar(label="Class")
plt.tight_layout()
plt.show()
````

# t-SNE (t-distributed Stochastic Neighbor Embedding)

t-SNE ist ein nicht-linearer Dimensionalitätsreduktionsalgorithmus, der
vor allem für die **Visualisierung von hochdimensionalen Daten in 2D oder 3D** verwendet wird.

---

#### Grundidee

- t-SNE versucht, **die lokale Struktur der Daten** beizubehalten:
  Punkte, die im hohen Raum nah beieinander liegen, sollen auch in der Einbettung nahe sein.
- Globale Distanzen werden weniger genau abgebildet, der Fokus liegt auf **lokalen Clustern**.

---

#### Wichtige Parameter

1. **n_components**: Anzahl der Dimensionen der Einbettung (typischerweise 2 oder 3 für Visualisierung).  
2. **perplexity**: Effektive Anzahl der Nachbarn, die bei der lokalen Struktur berücksichtigt werden.  
   - Kleine Werte → sehr lokale Struktur (wenige Nachbarn)  
   - Große Werte → größere Nachbarschaften, globalere Struktur  
   - Typische Werte: 5–50  
3. **learning_rate**: Lernrate für die Optimierung der Einbettung. `'auto'` passt oft gut.  
4. **init**: Initialisierung der Punkte (`'random'` oder `'pca'`).

---

#### Output

- **X_embedded**: Die niedrigdimensionale Einbettung der Datenpunkte.  
- Kann für **Scatterplots mit Klassenfarben** genutzt werden, um Cluster sichtbar zu machen.
