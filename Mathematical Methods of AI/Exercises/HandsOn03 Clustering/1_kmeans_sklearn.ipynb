```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
```

In this notebook, we will do the same as it was done in the previous, with the `sklearn` function for `KMeans`

```python
X, truth = make_blobs(
    n_samples=200,
    centers = [[0, 0], [1, 1], [2, 0]],
    cluster_std=0.2,
    random_state=42
)

plt.scatter(X[:, 0], X[:, 1], c=truth, cmap='viridis')
plt.show()
```

Run the following cell multiple times. Note that (i) since the method is unsupervised, the labels returned are not necessarily matching the order of the original ones and (ii) multiple runs will result in different models.

```python
# Inser here code to fit the model and plot the results
from sklearn.cluster import KMeans
K =3
K_Means_model = KMeans(n_clusters=K, random_state=None, n_init=10, max_iter = 100, tol=1e-4)
fit_K_means = K_Means_model.fit(X)
print(fit_K_means.cluster_centers_)
print(fit_K_means)
plt.scatter(X[:, 0], X[:, 1], c=fit_K_means.labels_, cmap='viridis')
plt.scatter(fit_K_means.cluster_centers_[:, 0], fit_K_means.cluster_centers_[:, 1], s=300, c='red', marker='X')
plt.show()

K_values = [2,3,4,5,6,7,8,9,10,15,20,30,40,50]
n_reps = 10
losses = np.zeros((n_reps, len(K_values)))

for idx_K,K in enumerate(K_values):
    for n in range(n_reps):
        # Fit model, store loss
        model_fit = KMeans(n_clusters=K, n_init=1).fit(X)
        losses[n, idx_K] = model_fit.inertia_
        pass


# Plot...
plt.plot(K_values, losses.T, 'o', alpha=0.5)
```

The `random_state` argument exists to guarantee reproducibility (see https://scikit-learn.org/stable/common_pitfalls.html#randomness)

If we fix random state one observes, the losses per inititialization of starting cluster centers do not differ anymore

```python
# As above, but include `random_state`
# Inser here code to fit the model and plot the results
from sklearn.cluster import KMeans
K =3
K_Means_model = KMeans(n_clusters=K, random_state=None, n_init=10, max_iter = 100, tol=1e-4)
fit_K_means = K_Means_model.fit(X)
print(fit_K_means.cluster_centers_)
print(fit_K_means)
plt.scatter(X[:, 0], X[:, 1], c=fit_K_means.labels_, cmap='viridis')
plt.scatter(fit_K_means.cluster_centers_[:, 0], fit_K_means.cluster_centers_[:, 1], s=300, c='red', marker='X')
plt.show()

K_values = [2,3,4,5,6,7,8,9,10,15,20,30,40,50]
n_reps = 10
losses = np.zeros((n_reps, len(K_values)))

for idx_K,K in enumerate(K_values):
    for n in range(n_reps):
        # Fit model, store loss
        model_fit = KMeans(n_clusters=K, n_init=1, random_state=42).fit(X)
        losses[n, idx_K] = model_fit.inertia_
        pass


# Plot...
plt.plot(K_values, losses.T, 'o', alpha=0.5)
```
