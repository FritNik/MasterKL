```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
```

In this notebook, other clustering methods offered by scikit-learn are explored. 
Those methods are not necessarily explained during the lectures, but feel free to explore them! A good starting point is [this comparison](https://scikit-learn.org/stable/auto_examples/cluster/plot_cluster_comparison.html#sphx-glr-auto-examples-cluster-plot-cluster-comparison-py)

Try changing the dataset!

```python
X, truth = make_blobs(
    n_samples=200,
    centers = [[0, 0], [1, 1], [2, 0]],
    cluster_std=0.2
)

plt.scatter(X[:, 0], X[:, 1], c=truth, cmap='viridis')
plt.show()
```

```python
from sklearn.cluster import DBSCAN

model = DBSCAN(eps=0.3, min_samples=5)
model.fit(X)
yhat = model.labels_

plt.scatter(X[:,0], X[:,1], c=truth, s=100, label='True labels')
plt.scatter(X[:, 0], X[:, 1], c=yhat, cmap='viridis')
plt.legend()
plt.show()
```

```python
from sklearn.mixture import GaussianMixture

model = GaussianMixture(n_components=3)
model.fit(X)
yhat = model.predict(X)

plt.scatter(X[:,0], X[:,1], c=truth, s=100, label='True labels')
plt.scatter(X[:, 0], X[:, 1], c=yhat, cmap='viridis')
plt.legend()
plt.show()
```

```python
from sklearn.cluster import SpectralClustering

model = SpectralClustering(n_components=3)
model.fit(X)
yhat = model.labels_

plt.scatter(X[:,0], X[:,1], c=truth, s=100, label='True labels')
plt.scatter(X[:, 0], X[:, 1], c=yhat, cmap='viridis')
plt.legend()
plt.show()
```
