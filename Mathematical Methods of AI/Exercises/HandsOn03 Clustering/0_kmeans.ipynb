Here we will write a scikit-learn compatible K-means implementation.

The necessary methods are:
- `__init__` - initialization for the hyperparameters. In this case, you will need the number of clusters, the maximal number of iterations, the tolerance, a random state and the number of random initializations.
- `fit` - fitting mechanism. For each 
- `predict` - prediction mechanism

For more information, see https://scikit-learn.org/stable/developers/develop.html


```python
class KMeansDemo:
    def __init__(self, random_state=None, K=3, max_iters=100, tol = 1e-4, n_init = 10):
        #... TODO: add the other parameters...
        self.random_state = np.random.RandomState(random_state)
        self.K = K
        self.max_iters = max_iters
        self.tol = tol
        self.n_init = n_init

    def _init_centroids(self,X):
        centroids = self.random_state.choice(X.shape[0], size=self.K, replace=False)
        return X[centroids].copy()
    
    def fit(self,X):

        n_samples = X.shape[0]
        best_loss = np.inf

        # Loop
        for _ in range(self.n_init):
            C = self._init_centroids(X)
            history = []

            for it in range(self.max_iters):
                
                # you can use `cdist` to compute distances and `np.argmin` to select closest.
                # ...
                distances = cdist(X, C, metric='sqeuclidean')
                gamma = np.zeros((n_samples, self.K))
                closest_indices = np.argmin(distances, axis=1)
                gamma[np.arange(n_samples), closest_indices] = 1 # T kreuz K Matrix

                # Keep track of the loss by summing the distances for the closest indices
                loss = np.sum(distances[np.arange(n_samples), closest_indices])
                history.append(loss)

                # Sufficient decrease, stop the loop
                if it > 0 and history[-2] - loss < self.tol:
                    break

                # Update C with the mean value of the points with respective affiliations
                C = (gamma.T @ X) / gamma.sum(axis=0)[:, None]
            
            final_loss = history[-1]

            # Now, check if the current iteration has better loss than the previous best
            if final_loss < best_loss:
                best_loss   = final_loss
                best_centroids = C.copy()
                best_gamma    = gamma.copy()
                best_history   = history.copy() 

        self.centroids_   = best_centroids
        self.gamma_      = best_gamma
        self.loss_history_= best_history
        self.n_iter_      = len(best_history)
        self.n_init_      = self.n_init
        return self
    
    def predict(self, X):
        dists  = cdist(X, self.centroids_, metric='sqeuclidean')
        return np.argmin(dists, axis=1)
```

```python
kmeans = KMeansDemo(K=3,
                    max_iters=200,
                    tol=1e-6,
                    n_init=10,          # try 10 initialisations
                    random_state=123)
kmeans.fit(X)

print(f"Best run used {kmeans.n_iter_} iterations")
print("Final loss (mean squared error):", kmeans.loss_history_[-1])
print("\nCentroids of the best model:")
print(kmeans.centroids_)
```

```python
plt.figure(figsize=(6,5))
scatter = plt.scatter(X[:,0], X[:,1],
                      c=kmeans.gamma_, cmap='viridis',
                      s=30, alpha=0.7)
plt.scatter(kmeans.centroids_[:,0], kmeans.centroids_[:,1],
            color='red', marker='X', s=200, linewidths=3)
plt.title('K‑Means (best of 10 starts)')
plt.xlabel('x₁')
plt.ylabel('x₂')
plt.legend(*scatter.legend_elements(), title="Cluster")
plt.show()
```
