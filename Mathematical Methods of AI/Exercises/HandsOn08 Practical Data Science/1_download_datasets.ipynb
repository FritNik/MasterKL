# How to obtain a dataset

## Synthetic dataset
For debugging, testing and assessment of specific properties of a model, it can be useful to create synthetic datasets. Scikit-learn provides some functions to create such datasets with different properties. Some of those were used in previous notebooks, here we demonstrate the use of some of them. A full list can be found in the [sklearn documentation](https://scikit-learn.org/stable/datasets/sample_generators.html).

## Classification

```python
from sklearn.datasets import make_blobs, make_moons, make_circles, make_classification

n_samples = 500 
X,y = make_blobs(n_samples=n_samples, random_state=42)
plt.scatter(X[:,0], X[:,1], c=y, edgecolors='k')
```

```python
X,y = make_moons(n_samples=n_samples, noise=0.1, random_state=42)
plt.scatter(X[:,0], X[:,1], c=y, edgecolors='k')
```

```python
X,y = make_circles(n_samples=n_samples, noise=0.1, factor=0.1, random_state=42)
plt.scatter(X[:,0], X[:,1], c=y, edgecolors='k')
```

```python
X,y = make_classification(
    n_samples=n_samples, 
    n_features=2,
    n_informative=1,
    n_redundant=0,
    n_classes=2,
    n_clusters_per_class=1,
    random_state=42)
plt.scatter(X[:,0], X[:,1], c=y, edgecolors='k')
```

## Real world datasets
Other than synthetic datasets, many datasets are available online. Some of them are directly available in sklearn, others can be downloaded from public repositories. A good list of public datasets can be found at the [UCI Machine Learning Repository](https://archive.ics.uci.edu/ml/index.php), on [Kaggle](https://www.kaggle.com/datasets) or on [OpenML](https://www.openml.org/search?type=data).

Scikit learn provides some datasets directly in the library, such as the Iris dataset. Other datasets can be downloaded using dedicated functions, such as the [20 newsgroups dataset](https://scikit-learn.org/stable/datasets/real_world.html#newsgroups-dataset). We will see some examples in the next cells, but for more details please refer to the [sklearn documentation](https://scikit-learn.org/stable/datasets/).

In general, you can either get the data as numpy arrays or as pandas DataFrames. The latter is often more convenient for data analysis, as it provides more information on the features and allows to easily manipulate the data.

```python
from sklearn.datasets import load_diabetes

dataset = load_diabetes(
    as_frame=True
)

dataset.data
```

## Other modalities
The datasets are not only tabular data, but can also be images, text, time series, graphs, etc. Scikit-learn provides some functions to load such datasets, but for more complex datasets other libraries are often used, such as [TensorFlow Datasets](https://www.tensorflow.org/datasets) or [PyTorch Datasets](https://pytorch.org/vision/stable/datasets.html).

```python
from sklearn.datasets import load_digits

X,y = load_digits(as_frame=False, return_X_y=True)

print(f"X is a {type(X)}, containing {X.shape[0]} examples with {X.shape[1]}-dimensional flattened images")

idx_to_check = 42
plt.imshow(X[idx_to_check,:].reshape(8,8))
plt.title(f"The {idx_to_check} image is a {y[idx_to_check]}")
plt.show()
```

```python
from sklearn.datasets import fetch_olivetti_faces

dataset = fetch_olivetti_faces()

idx_to_check = 10
plt.imshow(dataset.images[idx_to_check])
plt.show()
```
