```python
import numpy as np
import matplotlib.pyplot as plt
from time import perf_counter
```

```python
def find_coefficients(X,y):
    ones = np.ones((X.shape[0], 1))
    X_new = np.hstack((ones, X)) 
    thetas = np.linalg.inv(X_new.T @ X_new) @ X_new.T @ y
    return thetas
```

```python
# Fill this is:
#   - generate a range of D and T to use for data generation (check `linspace` and `logspace` functions)
#   - using loops, create data and fill the following array
#   - try fitting multiple times and taking the average time

Ds =np.logspace(0,2,5,dtype=int)  # e.g., dimensions from 1 to 100
Ts =np.logspace(1,4,5,dtype=int)  # e.g

duration = np.zeros((len(Ds), len(Ts)))

# you can use the following to take the time for a single evaluation:
        # start_time = perf_counter()
        # ... do something that takes time...
        # duration[i,j] = perf_counter() - start_time

for id, D in enumerate(Ds):
    for it, T in enumerate(Ts):

        X = np.random.rand(T,D)
        y = np.random.rand(T)

        start_time = perf_counter()
        for _ in range(10):
            coeffs =  find_coefficients(X,y)
        duration[id,it] = (perf_counter() - start_time)/10
```

```python
# The following code plots a surface, if everything above is done as suggested
fig, ax = plt.subplots(subplot_kw={'projection': '3d'})

dd,tt = np.meshgrid(Ds, Ts)

ax.plot_surface(dd,tt,duration.T)
ax.set_xlabel('D')
ax.set_ylabel('T')
ax.set_zlabel('Time (s)')
plt.tight_layout()
plt.show()
```
