```python
import numpy as np
from sklearn.datasets import make_classification
import matplotlib.pyplot as plt
from sklearn.tree import plot_tree

X,y = make_classification(
    n_features=2,
    n_informative=2,
    n_redundant=0,
    n_repeated=0,
    n_samples=200,
    class_sep=2.5,
    random_state=42
)

plt.scatter(X[:,0], X[:,1], c=y)
```

THe `RanfomForestClassifier` object has more parameters than a `DecisionTreeClassifier`, as you can control, between others:
- number of trees to be used
- whether to use boostrap aggregation
- how many samples per tree should be used
- how many features per tree should be provided
- hyperparameters for the trees

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    n_estimators=5,            # number of trees
    max_depth=4,                # depth of each tree
    bootstrap=True,             # whether to use bagging
    max_samples=50,             # number of samples per tree
    max_features=2,             # features to give to each tree
    min_samples_split=5,        
    random_state=42,
    criterion='entropy',
    n_jobs=-1                   # use all cores
)
rf.fit(X, y)

y_pred = rf.predict(X)

plt.scatter(X[:,0], X[:,1], c=y, s=150, edgecolors='k')
plt.scatter(X[:,0], X[:,1], c=y_pred, s=20, edgecolor='k')

print(f"The model contains {len(rf.estimators_)} elements of type {type(rf.estimators_[0])}")
```

```python
for i,tree in enumerate(rf.estimators_):
    plt.figure(figsize=(8,6))
    plot_tree(
        tree,
        feature_names=["d1", "d2"],
        class_names=["c1", "c2"],
        filled=True,
        rounded=True)
    plt.title(f"Tree number {i}")
    plt.tight_layout()
    plt.show()
```
