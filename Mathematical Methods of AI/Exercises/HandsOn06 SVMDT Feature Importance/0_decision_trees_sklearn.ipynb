This notebook demonstrates the use of Decision Trees with `sklearn`.
First, `DecisionTreeClassifier` will be used for a classification problem, and then `DecisionTreeRegressor` for a regression problem.

```python
import numpy as np
from sklearn.datasets import make_classification
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
```

We can use the `make_classification` function to generate a classification problem.
You can make it more difficult by, for example:
- decreasing the `class_sep`
- adding more features (`n_features`)
- inserting redundant or repeated features/reducing the informative features

```python
X,y = make_classification(
    n_features=2,
    n_informative=2,
    n_redundant=0,
    n_repeated=0,
    n_samples=200,
    class_sep=2.5,
    random_state=42
)

plt.scatter(X[:,0], X[:,1], c=y)
```

Fit a `DecisionTreeClassifier` object. Read the documentation for the possible options.

```python
clf = DecisionTreeClassifier(max_depth=4,
                             min_samples_split=5,
                             random_state=42,
                             criterion="entropy")
clf.fit(X, y)
```

You can obtain the predicted label for each point by using `clf.predict()` function.

```python
y_pred = clf.predict(X)

plt.scatter(X[:,0], X[:,1], c=y, s=150, edgecolors='k')
plt.scatter(X[:,0], X[:,1], c=y_pred, s=20, edgecolor='k')
```

THe actual decision tree can be visualized using the `plot_tree` function.

```python
from sklearn.tree import plot_tree

plt.figure(figsize=(12,10))
plot_tree(
    clf,
    feature_names=["d1", "d2"],
    class_names=["c1", "c2"],
    filled=True,
    rounded=True)
plt.show()
```

# Regression problem
As for the classification, we can use the `make_regression` function to generate a synthetic example

```python
from sklearn.datasets import make_regression
from sklearn.tree import DecisionTreeRegressor

X,y = make_regression(n_features=2, n_samples=200)
```

This time, we use a `DecisionTreeRegressor` object.
Note: with `max_depth=2`, the model does not have sufficient capacity to fit the data

```python
reg = DecisionTreeRegressor(max_depth=2).fit(X,y)
y_hat = reg.predict(X)

plt.scatter(X[:,0], X[:,1], c=y, s=150, edgecolors='k')
plt.scatter(X[:,0], X[:,1], c=y_hat, s=20, edgecolor='k')
```

```python
plt.figure(figsize=(12,10))
plot_tree(reg,feature_names=["d1", "d2"],
              filled=True,
              rounded=True)
plt.show()
```
