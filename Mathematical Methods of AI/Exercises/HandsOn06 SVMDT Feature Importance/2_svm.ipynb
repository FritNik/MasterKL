```python
import numpy as np
from sklearn.datasets import make_classification
import matplotlib.pyplot as plt

X,y = make_classification(
    n_features=2,
    n_informative=2,
    n_redundant=0,
    n_repeated=0,
    n_samples=200,
    class_sep=2.5,
    random_state=42
)
```

```python
from sklearn.svm import SVC

svm = SVC(
    kernel="linear",
).fit(X,y)

y_pred = svm.predict(X)

plt.scatter(X[:,0], X[:,1], c=y, s=150, edgecolors='k')
plt.scatter(X[:,0], X[:,1], c=y_pred, s=20, edgecolor='k')
```

The support vectors can be accessed after fitting in the following way:

```python
supp_vect = svm.support_vectors_

d1,d2 = 0,1 # if you have more than 2 dimensions, you can select which ones to plot here
plt.scatter(X[:,d1], X[:,d2], c=y, s=150, edgecolors='k')
plt.scatter(X[:,d1], X[:,d2], c=y_pred, s=20, edgecolor='k')
plt.scatter(supp_vect[:,d1], supp_vect[:,d2], c='r')
```
