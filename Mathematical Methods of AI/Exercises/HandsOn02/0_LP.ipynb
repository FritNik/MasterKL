```python
from scipy.optimize import linprog
import numpy as np
```

In this notebook, we will solve a Linear Programming problem using `scipy.optimize`'s `linprog`.

We need to express the problem in matrix form, using `A`, `b`, `c` and `bounds` (to have positive values only).

```python
# c = ...

# A = ...
# b = ...

bounds = [(0, np.inf), (0, np.inf)]
```

And then, we can call `linprog`. For information about it, see the documentation: https://docs.scipy.org/doc/scipy/reference/generated/scipy.optimize.linprog.html

```python
res = linprog(c, A_ub=A, b_ub=b, method='highs', bounds=bounds)
print(res)
```

The optimal values can be accessed as:

```python
res.x
```

You can impose integrality constraints using the `integrality` argument. You will need a vector with same number of elements as variables.

```python
res = linprog(c, A_ub=A, b_ub=b, method='highs', bounds=bounds, integrality=[1, 1])
print(res.x)
```
