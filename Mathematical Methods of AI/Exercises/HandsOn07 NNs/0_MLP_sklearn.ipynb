```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.neural_network import MLPClassifier
import matplotlib.pyplot as plt
```

The simplest way to fit a neural network is using the `MLPClassifier` or `MLPRegressor` functions from sklearn.
They are appropriate for supervised learning and convenient to use.

```python
iris = load_iris(as_frame=False)
X = np.array(iris.data)
y = iris.target
```

```python
# fit a MLPClassifier
classifier = MLPClassifier(hidden_layer_sizes=(10, 10), max_iter=1000, random_state=42)
classifier.fit(X,y)
```

After fitting, you can check the loss value during training by accessing `loss_curve_`

```python
plt.plot(classifier.loss_curve_)
plt.xlabel("Iteration")
plt.ylabel(r"$\mathcal{L}$")
```

As for other scikit-learn estimators, you can call `predict`, `score`, etc.

```python
# Plot the predictions...
print(classifier.score(X,y))
y_pred = classifier.predict(X)
plt.figure()
plt.scatter(X[:, 0], X[:, 1], c=y_pred)
```

Note: the above plot can be misleading, because the dataset has 4 dimensions. Do not forget to inspect different combinations

```python
preds = classifier.predict_proba(X)

fig, axs = plt.subplots(4,3)

for i in range(3):
    cc = axs[0,i].scatter(X[:,0], X[:,1], c=preds[:,i])
    axs[0,i].set_xlabel(r"$D_0$")
    plt.colorbar(cc)

    cc = axs[1,i].scatter(X[:,0], X[:,2], c=preds[:,i])
    axs[1,i].set_xlabel(r"$D_0$")
    plt.colorbar(cc)

    cc = axs[2,i].scatter(X[:,0], X[:,3], c=preds[:,i])
    axs[2,i].set_xlabel(r"$D_0$")
    plt.colorbar(cc)

    cc = axs[3,i].scatter(X[:,1], X[:,2], c=preds[:,i])
    axs[3,i].set_xlabel(r"$D_1$")
    plt.colorbar(cc)


axs[0,0].set_title("P(Class = 0)")
axs[0,1].set_title("P(Class = 1)")
axs[0,2].set_title("P(Class = 2)")

axs[0,0].set_ylabel(r"$D_1$")
axs[1,0].set_ylabel(r"$D_2$")
axs[2,0].set_ylabel(r"$D_3$")
axs[3,0].set_ylabel(r"$D_2$")

plt.tight_layout()
plt.show()
```
