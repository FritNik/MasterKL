In this notebook, we visualize some commonly used activation functions

```python
from torch.nn.functional import relu, sigmoid,tanh, softplus, elu, selu, celu, leaky_relu, mish
import matplotlib.pyplot as plt
import torch
```

```python
x = torch.linspace(-5, 5, 100)
y = mish(x)
y.numpy()
```

```python
acts = [sigmoid, relu, tanh, softplus, elu, selu, celu, leaky_relu, mish]
labels = ['sigmoid', 'relu', 'tanh', 'softplus', 'elu', 'selu', 'celu', 'leaky_relu', 'mish']

fig, axs = plt.subplots(3,3, figsize=(12, 12))
ax = axs.ravel()

for i, (label, act) in enumerate(zip(labels, acts)):
    x = torch.linspace(-5, 5, 100)
    y = act(x)
    ax[i].plot(x.numpy(), y.numpy())
    ax[i].set_title(label)
    ax[i].set_xlabel('x')
    ax[i].set_ylabel('f(x)')

plt.tight_layout()
plt.show()
```
