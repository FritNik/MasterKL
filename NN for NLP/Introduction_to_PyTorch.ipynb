#### Operations

Most operations that exist in numpy, also exist in PyTorch. A full list of operations can be found in the [PyTorch documentation](https://pytorch.org/docs/stable/tensors.html#), but we will review the most important ones here.

The simplest operation is to add two tensors:

```python
x1 = torch.rand(2, 3)
x2 = torch.rand(2, 3)

x1 = torch.tensor([1, 2, 3])
x2 = torch.tensor([1, 1, 1])

## TODO: Sum up the tensors x1 and x2, and store the result in y
y = x1 + x2

print("X1", x1)
print("X2", x2)
print("Y", y)
```

Calling `x1 + x2` creates a new tensor containing the sum of the two inputs. However, we can also use in-place operations that are applied directly on the memory of a tensor. We therefore change the values of `x2` without the chance to re-accessing the values of `x2` before the operation. An example is shown below:

```python
x1 = torch.tensor([1, 2, 3])
x2 = torch.tensor([2, 3, 4])
print("X1 (before)", x1)
print("X2 (before)", x2)

x2.add_(x1)
print("X1 (after)", x1)
print("X2 (after)", x2)
```

In-place operations are usually marked with a underscore postfix (e.g. "add_" instead of "add").

Other commonly used operations include matrix multiplications, which are essential for neural networks. Quite often, we have an input vector $\mathbf{x}$, which is transformed using a learned weight matrix $\mathbf{W}$. There are multiple ways and functions to perform matrix multiplication, some of which we list below:

* `torch.matmul`: Performs the matrix product over two tensors, where the specific behavior depends on the dimensions. If both inputs are matrices (2-dimensional tensors), it performs the standard matrix product. For higher dimensional inputs, the function supports broadcasting (for details see the [documentation](https://pytorch.org/docs/stable/generated/torch.matmul.html?highlight=matmul#torch.matmul)). Can also be written as `a @ b`, similar to numpy. 
* `torch.mm`: Performs the matrix product over two matrices, but doesn't support broadcasting (see [documentation](https://pytorch.org/docs/stable/generated/torch.mm.html?highlight=torch%20mm#torch.mm))
* `torch.bmm`: Performs the matrix product with a support batch dimension. If the first tensor $T$ is of shape ($b\times n\times m$), and the second tensor $R$ ($b\times m\times p$), the output $O$ is of shape ($b\times n\times p$), and has been calculated by performing $b$ matrix multiplications of the submatrices of $T$ and $R$: $O_i = T_i @ R_i$
* `torch.einsum`: Performs matrix multiplications and more (i.e. sums of products) using the Einstein summation convention. Explanation of the Einstein sum can be found in assignment 1.

### The Power of PyTorch: Dynamic Computation Graph and Automatic Gradients

PyTorch stands out as a go-to choice in the realm of Deep Learning due to its remarkable ability to automatically calculate the derivatives or gradients of the functions we define. In the context of our Deep Learning projects, we often work with neural networks, essentially sophisticated functions. These functions frequently involve weight matrices, which we refer to as the "parameters" or simply the "weights."

When our neural network yields a single scalar value, we commonly speak of taking its "derivative." However, you will notice that we often deal with scenarios where multiple output variables (or "values") are generated. In such cases, the term "gradients" is more appropriate, as it encompasses this broader concept.

Let's consider an input represented as $\mathbf{x}$. Our function is shaped by manipulating this input, typically involving matrix multiplications with weight matrices and adding bias vectors. As we perform these operations on our input, a "computational graph" is automatically constructed. This graph elucidates the path from our input to the final output. Notably, PyTorch follows a "define-by-run" approach, enabling us to focus on our manipulations while PyTorch meticulously tracks and manages the construction of this dynamic computation graph.

In summary, our primary task is to compute the desired "output." Subsequently, we can effortlessly call upon PyTorch to compute the gradients for us.

> **Why do we seek gradients?** Consider that we've formulated a neural network tasked with computing an output, denoted as $y$, for a given input vector $\mathbf{x}$. We also define an "error measure" that quantifies the discrepancy between our network's prediction of $y$ for input $\mathbf{x}$ and the actual value we desire. Armed with this error measure, we leverage the gradients to iteratively "update" the weights $\mathbf{W}$ responsible for the output. By doing so, we gradually steer the network towards generating output that aligns more closely with our target.

Our journey commences with the essential step of specifying which tensors necessitate gradients. By default, newly created tensors do not require gradients.

```python
x = torch.ones((3,))
print(x.requires_grad)
```

We can change this for an existing tensor using the function `requires_grad_()` (underscore indicating that this is a in-place operation). Alternatively, when creating a tensor, you can pass the argument `requires_grad=True` to most initializers we have seen above.

```python
x.requires_grad_(True)
print(x.requires_grad)
```

In order to get familiar with the concept of a computation graph, we will create one for the following function:

$$y = \frac{1}{|x|}\sum_i \left[(x_i + 2)^2 + 3\right]$$

You could imagine that $x$ are our parameters, and we want to optimize (either maximize or minimize) the output $y$. For this, we want to obtain the gradients $\partial y / \partial \mathbf{x}$. For our example, we'll use $\mathbf{x}=[0,1,2]$ as our input.

```python
## TODO: Create a tensor with values [0.0, 1.0, 2.0] which is requires gradients
# Note that only float tensors can have gradients
x = torch.tensor([0., 1., 2.], dtype=torch.float32)
x.requires_grad_(True)
print("X", x)
```

Now let's build the computation graph step by step. You can combine multiple operations in a single line, but we will separate them here to get a better understanding of how each operation is added to the computation graph.

```python
## TODO: Implement the function above in four steps, each representing one single operation (add -> a, square -> b, add -> c, mean -> y)
a = x + 2
b = a**2
c = b + 3
y = torch.sum(c) / len(x)
print("Y", y)
```

Using the statements above, we have created a computation graph that looks similar to the figure below:

<center style="width: 100%"><img src="pytorch_computation_graph.svg" width="200px"></center>

We calculate $a$ based on the inputs $x$ and the constant $2$, $b$ is $a$ squared, and so on. The visualization is an abstraction of the dependencies between inputs and outputs of the operations we have applied.
Each node of the computation graph has automatically defined a function for calculating the gradients with respect to its inputs, `grad_fn`. You can see this when we printed the output tensor $y$. This is why the computation graph is usually visualized in the reverse direction (arrows point from the result to the inputs). We can perform backpropagation on the computation graph by calling the function `backward()` on the last output, which effectively calculates the gradients for each tensor that has the property `requires_grad=True`:

```python
y.backward()
```

`x.grad` will now contain the gradient $\partial y/ \partial \mathcal{x}$, and this gradient indicates how a change in $\mathbf{x}$ will affect output $y$ given the current input $\mathbf{x}=[0,1,2]$:

```python
print(x.grad)
```

We can also verify these gradients by hand. We will calculate the gradients using the chain rule, in the same way as PyTorch did it:

$$\frac{\partial y}{\partial x_i} = \frac{\partial y}{\partial c_i}\frac{\partial c_i}{\partial b_i}\frac{\partial b_i}{\partial a_i}\frac{\partial a_i}{\partial x_i}$$

Note that we have simplified this equation to index notation, and by using the fact that all operation besides the mean do not combine the elements in the tensor. The partial derivatives are:

$$
\frac{\partial a_i}{\partial x_i} = 1,\hspace{1cm}
\frac{\partial b_i}{\partial a_i} = 2\cdot a_i\hspace{1cm}
\frac{\partial c_i}{\partial b_i} = 1\hspace{1cm}
\frac{\partial y}{\partial c_i} = \frac{1}{3}
$$

Hence, with the input being $\mathbf{x}=[0,1,2]$, our gradients are $\partial y/\partial \mathbf{x}=[4/3,2,8/3]$. The previous code cell should have printed the same result.

### The data

PyTorch also provides a few functionalities to load the training and test data efficiently, summarized in the package `torch.utils.data`.

```python
import torch.utils.data as data
```

The data package defines two classes which are the standard interface for handling data in PyTorch: `data.Dataset`, and `data.DataLoader`. The dataset class provides an uniform interface to access the training/test data, while the data loader makes sure to efficiently load and stack the data points from the dataset into batches during training.

#### The dataset class

The dataset class summarizes the basic functionality of a dataset in a natural way. To define a dataset in PyTorch, we simply specify two functions: `__getitem__`, and `__len__`. The get-item function has to return the $i$-th data point in the dataset, while the len function returns the size of the dataset. For the document dataset dataset, we can define the dataset class as follows:

```python
class DocDataset(data.Dataset):
    def __init__(self):
        super().__init__()
        self.generate_document_data()

    def generate_document_data(self):
        from sklearn.datasets import fetch_20newsgroups
        from sklearn.feature_extraction.text import CountVectorizer
        categories = ['alt.atheism', 'soc.religion.christian', 'comp.graphics', 'sci.med']
        twenty_train = fetch_20newsgroups(subset='train', categories=categories, shuffle=True, random_state=42)
        count_vect = CountVectorizer()
        X = count_vect.fit_transform(twenty_train.data).toarray()

        self.data  = torch.tensor(X, dtype=torch.float32)        
        self.label = torch.tensor(twenty_train.target, dtype=torch.long)  
        self.features = self.data.shape[1]
        self.vocab = count_vect.vocabulary_

    def __len__(self):
        return self.data.shape[0]

    def __getitem__(self, idx):
        return self.data[idx], self.label[idx]
```

Let's try to create such a dataset and inspect it:

```python
dataset = DocDataset()
print("Size of dataset:", len(dataset))
print("Data point 0:", dataset[0])
# also lets print the vocab
# print(dataset.vocab) (uncomment at your own risk :p)
```

#### The data loader class

The class `torch.utils.data.DataLoader` represents a Python iterable over a dataset with support for automatic batching, multi-process data loading and many more features. The data loader communicates with the dataset using the function `__getitem__`, and stacks its outputs as tensors over the first dimension to form a batch.
In contrast to the dataset class, we usually don't have to define our own data loader class, but can just create an object of it with the dataset as input. Additionally, we can configure our data loader with the following input arguments (only a selection, see full list [here](https://pytorch.org/docs/stable/data.html#torch.utils.data.DataLoader)):

* `batch_size`: Number of samples to stack per batch
* `shuffle`: If True, the data is returned in a random order. This is important during training for introducing stochasticity. 
* `drop_last`: If True, the last batch is dropped in case it is smaller than the specified batch size. This occurs when the dataset size is not a multiple of the batch size. Only potentially helpful during training to keep a consistent batch size.

#### nn.Module

In PyTorch, a neural network is build up out of modules. Modules can contain other modules, and a neural network is considered to be a module itself as well. The basic template of a module is as follows:

```python
class MyModule(nn.Module):
    def __init__(self):
        super().__init__()
    def forward(self, x):
        return x
```


The `forward` function serves as the core of a module, where the actual computation takes place. It gets executed when you call the module, typically in the form of `nn = MyModule(); nn(x)`. On the other hand, in the `init` function, we typically set up the module's parameters using `nn.Parameter` or define other modules that will be utilized within the `forward` function. It's important to note that the backward calculation, which handles gradients and backpropagation, is automatically managed by PyTorch, but it can also be customized if needed.

#### Simple Classifier

Now, armed with the pre-defined modules from the `torch.nn` package, we can proceed to design our own compact neural network. In this example, we'll create a minimal network consisting of an input layer, a single hidden layer with a tanh activation function, and an output layer. In other words, our network architecture should resemble the following diagram:

<center width="100%"><img src="nn.png" width="300px"></center>

In this visualization, the blue nodes represent the input neurons, which correspond to the data point's features. The yellow nodes represent the hidden neurons, featuring a tanh activation function, and the red node represents the output neuron. In PyTorch, you can define this neural network as follows:

```python
import torch
import torch.nn as nn

class SimpleClassifier(nn.Module):
    def __init__(self, num_inputs, num_hidden, num_outputs):
        super().__init__()
        self.linear1 = nn.Linear(num_inputs, num_hidden)
        self.act_fn = nn.ReLU()
        self.linear2 = nn.Linear(num_hidden, num_outputs)

    def forward(self, x):
        x = self.linear1(x)    #[B,num_hidden]
        x = self.act_fn(x)     #[B,num_hidden]
        x = self.linear2(x)    #[B,Num_clases]
        return x
```

In the examples presented in this notebook, we'll employ a compact neural network featuring two input neurons and four hidden neurons. Since we're dealing with straightforward classification tasks, the output dimension of the network corresponds to the number of classes we are classifying into. It's worth noting that we do not apply a softmax activation function to the output at this stage. The rationale behind this choice is that for various computations, especially those related to loss functions, it is often more efficient and accurate to work with the original network outputs rather than applying softmax. We'll delve into the detailed explanation for this decision later in the tutorial.

```python
model = SimpleClassifier(num_inputs=dataset.features, num_hidden=256, num_outputs=4)
print(model)
```

Printing the model lists all submodules it contains. The parameters of a module can be obtained by using its `parameters()` functions, or `named_parameters()` to get a name to each parameter object. For our small neural network, we have the following parameters:

```python
for name, param in model.named_parameters():
    print("Parameter %s, shape %s" % (name, str(param.shape)))
```

Each linear layer has a weight matrix of the shape `[output, input]`, and a bias of the shape `[output]`. The tanh activation function does not have any parameters. Note that parameters are only registered for `nn.Module` objects that are direct object attributes, i.e. `self.a = ...`. If you define a list of modules, the parameters of those are not registered for the outer module and can cause some issues when you try to optimize your module. There are alternatives, like `nn.ModuleList`, `nn.ModuleDict` and `nn.Sequential`, that allow you to have different data structures of modules. We will use them in a few later tutorials and explain them there. 

### Model Optimization Workflow

Once the model and dataset are defined, the next step is to set up the optimization process for the model. During the training phase, the following key steps are executed:

1. **Data Batch Retrieval:** Fetch a batch of data from the data loader.
2. **Model Predictions:** Obtain predictions from the model for the batch.
3. **Loss Calculation:** Compute the loss by assessing the disparity between predictions and actual labels.
4. **Backpropagation:** Calculate the gradients for each model parameter with respect to the loss.
5. **Parameter Updates:** Update the model's parameters by moving them in the direction indicated by the gradients.

Up to this point, we have discussed how to perform steps 1, 2, and 4 using PyTorch. Now, we will delve into the details of steps 3 and 5.

#### Loss Computation Modules

To calculate the loss for a batch, we can leverage the power of tensor operations, as these are seamlessly integrated into the computation graph. For instance, in the context of multi-class classification, the Cross Entropy (CE) loss is a commonly used metric, defined as:

$$\mathcal{L}_{CE} = \sum_i y_i \log x_i $$

Here, $y$ represents the labels, and $x$ signifies our model's predictions. However, PyTorch offers a selection of pre-defined loss functions for our convenience (refer to the full list [here](https://pytorch.org/docs/stable/nn.html#loss-functions)). In the case of CE, PyTorch provides `nn.CrossEntropyLoss`, which is recommended for its numerical stability. This is due to the softmax operation applied within the loss function. Therefore, it is advisable to employ loss functions that operate on "logits" whenever possible (without applying a softmax to the model's output in this scenario). In the context of the model defined earlier, we opt to utilize the `nn.CrossEntropyLoss` module. Please read the documentations of `nn.CrossEntropyLoss` [here](https://pytorch.org/docs/stable/generated/torch.nn.CrossEntropyLoss.html).

```python
loss_module = nn.CrossEntropyLoss()
```

#### Stochastic Gradient Descent (SGD)

In the context of parameter updates, PyTorch offers the `torch.optim` package, which incorporates a variety of widely-used optimization algorithms. Detailed discussions about specific optimizers and their distinctions will follow later in the course. For now, we will employ one of the simplest optimization techniques: `torch.optim.SGD`. Stochastic Gradient Descent operates by scaling the gradients with a small constant known as the learning rate and then subtracting this scaled value from the model's parameters, effectively reducing the loss. As a result, this process gradually steers the model towards minimizing the loss function. A suitable default learning rate for a modest-sized network, such as the one we have, is typically set at 0.001.

```python
ptimizer = torch.optim.SGD(model.parameters(), lr=0.001)
```

The optimizer provides two essential functions: `optimizer.step()` and `optimizer.zero_grad()`. The `step` function handles parameter updates based on the previously computed gradients, as described earlier. On the other hand, the function `optimizer.zero_grad()` plays a crucial role, especially prior to backpropagation. While it might seem less significant at first glance, it is a vital preparatory step. If we were to invoke the `backward` function on the loss when parameter gradients still retain non-zero values from the previous batch, the new gradients would be added to the existing ones rather than overwriting them.

This behavior occurs because a parameter can appear multiple times in a computation graph, and in such cases, we need to sum the gradients instead of replacing them. Therefore, it's imperative to remember to call `optimizer.zero_grad()` to reset the gradients to zero before calculating the gradients for a new batch.

Now, let's construct a concise training function. Recall the five key steps in our training process: load a batch, acquire predictions, calculate the loss, perform backpropagation, and update the model's parameters. In addition to these steps, we must also ensure that all data and model parameters are transferred to the device of our choice, typically the GPU if available.

It's important to note that for our small neural network, the data transfer to the GPU consumes more time than the potential performance gain from GPU acceleration. In cases involving larger networks, the time spent on data communication becomes significantly smaller relative to the overall runtime, making a GPU an essential asset. However, for the sake of practice, we will push the data to the GPU in this example.

```python
# Push model to device. Has to be only done once
device = "cuda"
model.to(device)
```

In addition, we set our model to training mode. This is done by calling `model.train()`. There exist certain modules that need to perform a different forward step during training than during testing (e.g. BatchNorm and Dropout), and we can switch between them using `model.train()` and `model.eval()`.

```python
def train_model(model, optimizer, data_loader, loss_module, num_epochs=100):
    model.train() 
 
    for epoch in tqdm(range(num_epochs), desc="Epoch"):
        for i, (data_inputs, data_labels) in enumerate(data_loader):
            data_inputs = data_inputs.to(device)
            data_labels = data_labels.to(device)
            preds = model(data_inputs)            
            loss = loss_module(preds, data_labels)
            optimizer.zero_grad()
            loss.backward()
            optimizer.step()

            if i % 100 == 0:
                print(f"Epoch: {epoch}, Iteration: {i}, Loss: {loss.item():.4f}")
    return model, optimizer
```

### Evaluation

Once we have trained a model, it is time to evaluate it on a held-out test set. As our dataset consist of randomly generated data points, we need to first create a test set with a corresponding data loader.

```python
test_data_loader = data.DataLoader(test_dataset, batch_size=128, shuffle=True)
```

We will employ accuracy as our metric, computed using the following formula:

$$\text{accuracy} = \frac{\text{number of correct predictions}}{\text{total number of predictions}}$$

When assessing the model, there's no need to maintain the computational graph since we won't be performing gradient calculations. This optimization reduces memory usage and enhances model speed. In PyTorch, you can disable the computation graph with the `torch.no_grad()` context manager. Additionally, don't forget to switch the model to evaluation mode.

```python
def eval_model(model, data_loader):
    model.eval() 
    device = next(model.parameters()).device  
    true_preds, num_preds = 0, 0

    with torch.no_grad():  
        for data_inputs, data_labels in data_loader:
            data_inputs = data_inputs.to(device)
            if not torch.is_floating_point(data_inputs):
                data_inputs = data_inputs.float()
            data_labels = data_labels.to(device).long()

            logits = model(data_inputs)

            if logits.ndim == 1 or (logits.ndim == 2 and logits.shape[1] == 1):
                pred_labels = (torch.sigmoid(logits.view(-1)) >= 0.5).long()
            else:
                pred_labels = logits.argmax(dim=1)

            true_preds += (pred_labels == data_labels).sum().item()
            num_preds  += data_labels.size(0)

    acc = true_preds / num_preds if num_preds else 0.0
    print(f"Accuracy of the model: {100.0 * acc:4.2f}%")
    return acc
```

```python
eval_model(model, test_data_loader)
```
